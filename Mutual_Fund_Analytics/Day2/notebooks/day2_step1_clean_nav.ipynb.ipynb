```python
import pandas as pd

df = pd.read_csv('C:/Users/deepu/desktop/Projects/02_nav_history (1).csv')

print(df.shape)
print(df.head())
print(df.dtypes)
```

```python
df['date'] = pd.to_datetime(df['date'], errors='coerce')

print("Unparseable dates:", df['date'].isna().sum())
```

```python
df = df.sort_values(by=['amfi_code', 'date']).reset_index(drop=True)
print(df.head(10))
```

```python
df['nav'] = df.groupby('amfi_code')['nav'].ffill()

print("Remaining nulls in nav:", df['nav'].isna().sum())
```

```python
before = df.shape[0]
df = df.drop_duplicates(subset=['amfi_code', 'date'], keep='first')
after = df.shape[0]

print(f"Dropped {before - after} duplicate rows")
```

```python
invalid_nav = df[df['nav'] <= 0]
print(f"Invalid NAV rows found: {len(invalid_nav)}")
print(invalid_nav)


df = df[df['nav'] > 0].reset_index(drop=True)
```

```python
df.to_csv('C:/Users/deepu/desktop/Projects/clean_nav.csv', index=False)

print("Saved clean_nav.csv with", df.shape[0], "rows")
```
