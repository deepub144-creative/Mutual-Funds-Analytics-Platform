```python
import pandas as pd

# 1. Load the CSV
df = pd.read_csv('07_scheme_performance (1).csv')

print(df.shape)
print(df.head())
```

```python
for col in ['return_1yr_pct', 'return_3yr_pct', 'return_5yr_pct']:
    df[col] = pd.to_numeric(df[col], errors='coerce')
    print(f"Non-numeric values in {col} (now NaN):", df[col].isna().sum())
```

```python
df['sharpe_flag'] = df['sharpe_ratio'] < 0

print("Negative Sharpe ratio count:", df['sharpe_flag'].sum())
print(df[df['sharpe_flag']][['scheme_name', 'sharpe_ratio']])
```

```python
df['expense_ratio_flag'] = ~df['expense_ratio_pct'].between(0.1, 2.5)

print("Out-of-range expense ratios:", df['expense_ratio_flag'].sum())
print(df[df['expense_ratio_flag']][['scheme_name', 'expense_ratio_pct']])
```
