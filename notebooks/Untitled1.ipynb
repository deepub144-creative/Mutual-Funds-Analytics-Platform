```python
import sqlite3
import pandas as pd

# Use absolute path
db_path = r"C:\Users\deepu\OneDrive\Desktop\bluestock_mf_capstone\data\db\bluestock_mf.db"
conn = sqlite3.connect(db_path)

print("🏆 Top 5 Fund Houses by Average AUM:")
query1 = """
SELECT fund_house, ROUND(AVG(aum_crore), 2) AS avg_aum_crore
FROM aum_by_fund_house
GROUP BY fund_house
ORDER BY avg_aum_crore DESC
LIMIT 5;
"""
display(pd.read_sql(query1, conn))

print("\n📊 Top 5 Categories by Net Inflow (Latest Month):")
query2 = """
WITH latest_month AS (SELECT MAX(month) AS max_month FROM category_inflows)
SELECT category, net_inflow_crore
FROM category_inflows
WHERE month = (SELECT max_month FROM latest_month)
ORDER BY net_inflow_crore DESC
LIMIT 5;
"""
display(pd.read_sql(query2, conn))

print("\n💰 Investor Transaction Summary:")
query3 = """
SELECT transaction_type, COUNT(*) AS total_transactions,
       ROUND(AVG(amount_inr), 2) AS avg_amount,
       ROUND(SUM(amount_inr), 2) AS total_amount
FROM investor_transactions
GROUP BY transaction_type;
"""
display(pd.read_sql(query3, conn))

print("\n📈 Folio Growth Over Time:")
query4 = """
SELECT month, total_folios_crore, equity_folios_crore, debt_folios_crore
FROM industry_folio_count
ORDER BY month;
"""
display(pd.read_sql(query4, conn))

conn.close()
print("✅ SQL queries completed.")
```
