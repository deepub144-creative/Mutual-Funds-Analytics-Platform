```python
import pandas as pd
fund_master = pd.read_csv("C:/Users/deepu/OneDrive/Desktop/intership/Day1/Mutual_fund_project/01_fund_master.csv")
nav_history = pd.read_csv("C:/Users/deepu/OneDrive/Desktop/intership/Day1/Mutual_fund_project/02_nav_history.csv")
print("Fund Master Columns:")
print(fund_master.columns.tolist())

print("\nNAV History Columns:")
print(nav_history.columns.tolist())

merged = fund_master.merge(
    nav_history,
    on="amfi_code",
    how="left",
    indicator=True
)

missing = merged[merged["_merge"] == "left_only"]

print("Total Fund Master records :", len(fund_master))
print("Matched records           :", (merged["_merge"] == "both").sum())
print("Missing in NAV History    :", len(missing))

print(missing["amfi_code"].tolist())


```
